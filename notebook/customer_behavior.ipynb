```python
#q1."Discount usage differs across purchasing categories."
sns.countplot(
    data=df,
    x="category",
    hue="discount_applied"
)

plt.title("Discount Usage Across Product Categories")
plt.xlabel("Category")
plt.ylabel("Number of Purchases")

plt.show()
```

```python
#q2.Which product categories generate the most revenue?
df1=df.groupby("category")["purchase_amount"].sum() \
      .sort_values()
ax=sns.barplot(x=df1.index,y=df1.values)
plt.bar_label(ax.containers[0], fmt="$%.0f")
plt.title("Total Revenue by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Revenue")
plt.tight_layout()
plt.show()
```

```python
#q3.How are different numerical customer behavior factors related to each other?
corr = df[
    ["age", "purchase_amount", "review_rating",
     "previous_purchases", "purchase_frequency_days"]
].corr()

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Customer Behavior Correlation Heatmap")
plt.show()
```

```python
#Q4. Does subscription status vary across different purchase frequencies?
plt.figure(figsize=(6,6))
sns.countplot(
    data=df,
    x="frequency_of_purchases",
    hue="subscription_status"
)


plt.title("Subscription Status Across Purchase Frequencies")
plt.xlabel("Purchase Frequency")
plt.ylabel("Number of Customers")
plt.yticks(range(0, 501, 50))
plt.tight_layout()
plt.show()
```

```python
#q5.Which product categories have the highest number of reviewed purchases?
data = df.groupby("category")["review_rating"].count().sort_values(ascending=False)

ax=sns.barplot(x=data.index, y=data.values,color='g')
plt.bar_label(ax.containers[0],fmt="%.0f")
plt.title("Number of Reviewed Purchases by Category")
plt.xlabel("Category")
plt.ylabel("Number of Reviews")
plt.show()
```
